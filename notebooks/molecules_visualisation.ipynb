# Visualisation of Molecule Data

```python
import numpy as np
import matplotlib.pyplot as plt
import os
import sys

# Add project root to path
sys.path.append('..')

from src.molecules import Molecules

molecules = Molecules(left_cut=630, right_cut=995)
```

## Check to reproduce Marcel's Plots

```python
fig, (ax, ax1, ax2) = plt.subplots(ncols=3,figsize=(15,5))
print("Cut wavelengths:", molecules.cut_wavelengths)

ax.plot(molecules.cut_wavelengths, molecules.molecules.hbo2_f, label='HbO_2')
ax.plot(molecules.cut_wavelengths, molecules.molecules.hb_f, label='Hbb')

ax1.plot(molecules.cut_wavelengths, molecules.molecules.coxa, label='oxCCO')
ax1.plot(molecules.cut_wavelengths, molecules.molecules.creda, label='redCCO')
ax1.plot(molecules.cut_wavelengths, molecules.molecules.c_oxy, label='oxCC')
ax1.plot(molecules.cut_wavelengths, molecules.molecules.c_red, label='redCC')
ax1.plot(molecules.cut_wavelengths, molecules.molecules.b_oxy, label='oxCB')
ax1.plot(molecules.cut_wavelengths, molecules.molecules.b_red, label='redCB')

ax2.plot(molecules.cut_wavelengths, molecules.molecules.fat, label='Fat')
ax2.plot(molecules.cut_wavelengths, molecules.molecules.water, label='Water')

ax.legend()
ax1.legend()
ax2.legend()
plt.title("Spectra")
```
